```python
from collections import Counter
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
from sklearn.metrics import calinski_harabasz_score
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

# Load data
data = pd.read_excel(r"E:\desktop\Multiobjective Identification of Phosphor Materials via a Domain Knowledge Integrated Semisupervised Learning Framework\dataset\stage4\296×87.xlsx")
X = data.iloc[:, 3:]
X
feature_names = X.columns
X = StandardScaler().fit_transform(X)
# Normalize data.
min_val = np.min(X, axis=0)
max_val = np.max(X, axis=0)
X = (X - min_val) / (max_val - min_val)
imputer = SimpleImputer(strategy="mean")
X = imputer.fit_transform(X)
X.shape
```

```python
# Get reported structure IDs
def extract_numbers(file_path):
    df = pd.read_excel(file_path)
    valid_numbers = []
    for index, value in df['composition'].items():
        if not pd.isna(value):
            valid_numbers.append(int(df.at[index, 'number']))
    return valid_numbers

file_path = r"E:\desktop\Multiobjective Identification of Phosphor Materials via a Domain Knowledge Integrated Semisupervised Learning Framework\dataset\stage4\benchmark4.xlsx"
reported_structures = extract_numbers(file_path)
print(reported_structures)

# Train a KNN model to reassign noise points generated by DBSCAN.
def remove_noise_by_nearest_cluster(X, labels):
    """Assign noise points to the cluster of the nearest non-noise points"""
    # Find all noise points
    noise_mask = labels == -1

    # Return directly if there are no noise points
    if not any(noise_mask):
        return labels

    # Find non-noise points
    core_samples = ~noise_mask
    X_core = X[core_samples]
    y_core = labels[core_samples]

    # Check whether non-noise points exist
    if len(X_core) == 0:
        # If all points are noise, assign them to a cluster randomly
        new_labels = np.zeros_like(labels)
        return new_labels

    # Train the KNN classifier
    nn = NearestNeighbors(n_neighbors=1).fit(X_core)
    distances, indices = nn.kneighbors(X[noise_mask])

    # Assign noise points to the nearest non-noise cluster
    new_labels = labels.copy()
    new_labels[noise_mask] = y_core[indices.flatten()]
    return new_labels

for j in range(30, 100, 5):
    j = j/100
    # Reduce dimensionality using PCA.
    pca = PCA(n_components= j)
    pca_data = pca.fit_transform(X)
    num_pca = pca_data.shape[1]

    for eps in range(1,100,1):
        eps = eps/100
        for min_samples in range(6, 21,3):
            db = DBSCAN(eps=eps, metric='cosine', min_samples=min_samples)
            db.fit(pca_data)
            labels = db.labels_
            labels = remove_noise_by_nearest_cluster(pca_data, labels)
            if 7 > len(np.unique(labels)) > 1:
                print(f"\nSilhouette score: {silhouette_score(pca_data, labels):.2f}", f"\nCalinski-Harabasz score: {calinski_harabasz_score(pca_data, labels):.2f}")
                data['cluster_label'] = labels
                unique_labels = np.unique(labels)
                print( f"\nPCA={j}",f"\neps={eps}", f"\nmin_samples={min_samples}", f"\nNumber of clusters={len(unique_labels)}")

                dict = {}    # Summarize each cluster.
                        # Summarize each cluster
                label_1 = []
                for num  in  reported_structures:
                        label = labels[num-1]
                        label_1.append(int(label))
                print(label_1)

                for label in unique_labels:
                    # All samples in the current cluster
                    cluster_samples = data[data['cluster_label'] == label]
                    total_in_cluster = len(cluster_samples)

                    # Number of reported samples
                    reported_in_cluster = len([num for num in cluster_samples['number']
                                            if num in reported_structures])

                    print(f"Cluster {label}: {reported_in_cluster}/{total_in_cluster} "
                        f"(Reported ratio: {reported_in_cluster/total_in_cluster:.2%})")

                # Print label distribution
                # print("Label distribution:", Counter(labels))

            else:
                    pass

```

```python
# ============================================================


#







#


# ============================================================

from pathlib import Path
from itertools import combinations

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import pairwise_distances


# ============================================================

# ============================================================

required_vars = ['data', 'reported_structures', 'pca_data', 'labels']

missing_vars = [
    v for v in required_vars
    if v not in globals()
]

if missing_vars:
    raise RuntimeError(
        f"Missing variables: {missing_vars}。"
        "Run the data-loading, benchmark-loading, and final clustering cells first."
    )

if len(reported_structures) != 15:
    raise ValueError(
        f"Current reported_structures contains "
        f"{len(reported_structures)} structures, "
        "instead of the expected 15."
    )

if len(pca_data) != len(data) or len(labels) != len(data):
    raise ValueError(
        "new_data / labels do not match the number of rows in data. "
        "Confirm that the final clustering result is being used."
    )


# ============================================================

# ============================================================

number_series = pd.to_numeric(
    data['number'],
    errors='coerce'
)

benchmark_row_indices = []

for num in reported_structures:

    hit = np.flatnonzero(
        number_series.to_numpy() == int(num)
    )

    if len(hit) == 0:
        raise ValueError(
            f"Cannot find "
            f"benchmark number={num}"
        )

    if len(hit) > 1:
        raise ValueError(
            f"benchmark number={num} "
            f"appears in data {len(hit)} times, "
            "so it cannot be uniquely located."
        )

    benchmark_row_indices.append(
        int(hit[0])
    )



benchmark_pca = np.asarray(pca_data)[
    benchmark_row_indices, :
]

benchmark_clusters = np.asarray(labels)[
    benchmark_row_indices
]

print(
    f"Benchmark PCA dimensions: "
    f"{benchmark_pca.shape[1]}"
)

print(
    f"Benchmark count: "
    f"{benchmark_pca.shape[0]}"
)


# ============================================================

# ============================================================

composition_map = {}

try:

    benchmark_meta = pd.read_excel(file_path)

    if (
        'number' in benchmark_meta.columns
        and 'composition' in benchmark_meta.columns
    ):

        tmp = benchmark_meta[
            ['number', 'composition']
        ].copy()

        tmp['number'] = pd.to_numeric(
            tmp['number'],
            errors='coerce'
        )

        tmp = (
            tmp
            .dropna(subset=['number'])
            .drop_duplicates('number')
        )

        composition_map = {
            int(row['number']): str(row['composition'])
            for _, row in tmp.iterrows()
            if not pd.isna(row['composition'])
        }

except Exception as e:

    print(
        "Note: composition information could not be read. "
        f"Only number will be used as the label. Reason: {e}"
    )


benchmark_compositions = [
    composition_map.get(
        int(n),
        f"No.{int(n)}"
    )
    for n in reported_structures
]


benchmark_labels = [
    f"{int(n)} | {c}"
    for n, c in zip(
        reported_structures,
        benchmark_compositions
    )
]


# ============================================================

# ============================================================

distance_matrix = pairwise_distances(
    benchmark_pca,
    metric='euclidean'
)


np.fill_diagonal(
    distance_matrix,
    0.0
)


distance_df = pd.DataFrame(
    distance_matrix,
    index=benchmark_labels,
    columns=benchmark_labels
)

distance_df.index.name = 'Benchmark'


# ============================================================

# ============================================================



#
# normalized_distance = distance / max(distance)
#



max_distance = np.max(distance_matrix)

if max_distance <= 0:
    raise ValueError(
        "The maximum Euclidean distance is zero, so normalization cannot be performed."
    )


normalized_distance_matrix = (
    distance_matrix / max_distance
)

np.fill_diagonal(
    normalized_distance_matrix,
    0.0
)


normalized_distance_df = pd.DataFrame(
    normalized_distance_matrix,
    index=benchmark_labels,
    columns=benchmark_labels
)

normalized_distance_df.index.name = 'Benchmark'


print(
    f"\nMaximum Euclidean distance: "
    f"{max_distance:.6f}"
)

print(
    "Normalized distance range: "
    f"{normalized_distance_matrix.min():.4f} "
    "– "
    f"{normalized_distance_matrix.max():.4f}"
)


# ============================================================

# ============================================================

pair_records = []

for i, j in combinations(
    range(len(reported_structures)),
    2
):

    pair_records.append({

        'Benchmark_A_Number':
            int(reported_structures[i]),

        'Benchmark_A_Composition':
            benchmark_compositions[i],

        'Cluster_A':
            int(benchmark_clusters[i]),

        'Benchmark_B_Number':
            int(reported_structures[j]),

        'Benchmark_B_Composition':
            benchmark_compositions[j],

        'Cluster_B':
            int(benchmark_clusters[j]),

        'Same_Cluster':
            bool(
                benchmark_clusters[i]
                == benchmark_clusters[j]
            ),


        'Euclidean_Distance':
            float(
                distance_matrix[i, j]
            ),


        'Normalized_Euclidean_Distance':
            float(
                normalized_distance_matrix[i, j]
            ),
    })


pairwise_df = pd.DataFrame(
    pair_records
)



pairwise_df = (
    pairwise_df
    .sort_values(
        'Euclidean_Distance',
        ascending=True
    )
    .reset_index(drop=True)
)


pairwise_df.insert(
    0,
    'Pair_Rank_From_Nearest',
    np.arange(
        1,
        len(pairwise_df) + 1
    )
)


# ============================================================

# ============================================================

pc_cols = [
    f'PC{i+1}'
    for i in range(
        benchmark_pca.shape[1]
    )
]


benchmark_info_df = pd.DataFrame(
    benchmark_pca,
    columns=pc_cols
)


benchmark_info_df.insert(
    0,
    'Cluster',
    benchmark_clusters.astype(int)
)


benchmark_info_df.insert(
    0,
    'Composition',
    benchmark_compositions
)


benchmark_info_df.insert(
    0,
    'Number',
    [
        int(x)
        for x in reported_structures
    ]
)


# ============================================================

# ============================================================

sort_order = np.argsort(
    benchmark_clusters,
    kind='stable'
)


sorted_labels = [
    benchmark_labels[k]
    for k in sort_order
]


# -------------------------

# -------------------------

sorted_distance_df = pd.DataFrame(

    distance_matrix[
        np.ix_(
            sort_order,
            sort_order
        )
    ],

    index=sorted_labels,
    columns=sorted_labels
)

sorted_distance_df.index.name = 'Benchmark'


# -------------------------

# -------------------------

sorted_normalized_distance_df = pd.DataFrame(

    normalized_distance_matrix[
        np.ix_(
            sort_order,
            sort_order
        )
    ],

    index=sorted_labels,
    columns=sorted_labels
)

sorted_normalized_distance_df.index.name = 'Benchmark'


# ============================================================

# ============================================================

same_cluster_raw = pairwise_df.loc[
    pairwise_df['Same_Cluster'],
    'Euclidean_Distance'
]

diff_cluster_raw = pairwise_df.loc[
    ~pairwise_df['Same_Cluster'],
    'Euclidean_Distance'
]


same_cluster_norm = pairwise_df.loc[
    pairwise_df['Same_Cluster'],
    'Normalized_Euclidean_Distance'
]

diff_cluster_norm = pairwise_df.loc[
    ~pairwise_df['Same_Cluster'],
    'Normalized_Euclidean_Distance'
]


summary_df = pd.DataFrame({

    'Metric': [

        'Number_of_benchmarks',

        'Number_of_unique_pairs',

        'Maximum_raw_distance',

        'Mean_raw_distance_all_pairs',

        'Median_raw_distance_all_pairs',

        'Mean_raw_distance_same_cluster',

        'Median_raw_distance_same_cluster',

        'Mean_raw_distance_different_cluster',

        'Median_raw_distance_different_cluster',

        'Mean_normalized_distance_all_pairs',

        'Median_normalized_distance_all_pairs',

        'Mean_normalized_distance_same_cluster',

        'Median_normalized_distance_same_cluster',

        'Mean_normalized_distance_different_cluster',

        'Median_normalized_distance_different_cluster',
    ],


    'Value': [

        len(reported_structures),

        len(pairwise_df),

        max_distance,

        pairwise_df[
            'Euclidean_Distance'
        ].mean(),

        pairwise_df[
            'Euclidean_Distance'
        ].median(),

        same_cluster_raw.mean()
        if len(same_cluster_raw)
        else np.nan,

        same_cluster_raw.median()
        if len(same_cluster_raw)
        else np.nan,

        diff_cluster_raw.mean()
        if len(diff_cluster_raw)
        else np.nan,

        diff_cluster_raw.median()
        if len(diff_cluster_raw)
        else np.nan,

        pairwise_df[
            'Normalized_Euclidean_Distance'
        ].mean(),

        pairwise_df[
            'Normalized_Euclidean_Distance'
        ].median(),

        same_cluster_norm.mean()
        if len(same_cluster_norm)
        else np.nan,

        same_cluster_norm.median()
        if len(same_cluster_norm)
        else np.nan,

        diff_cluster_norm.mean()
        if len(diff_cluster_norm)
        else np.nan,

        diff_cluster_norm.median()
        if len(diff_cluster_norm)
        else np.nan,
    ]
})


# ============================================================

# ============================================================

output_dir = (
    Path.cwd()
    / 'benchmark_distance_output'
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


excel_path = (
    output_dir
    / '15_benchmark_pairwise_euclidean_distance_normalized.xlsx'
)


with pd.ExcelWriter(
    excel_path,
    engine='openpyxl'
) as writer:


    distance_df.to_excel(
        writer,
        sheet_name='Distance_Matrix'
    )


    normalized_distance_df.to_excel(
        writer,
        sheet_name='Normalized_Distance_Matrix'
    )


    pairwise_df.to_excel(
        writer,
        sheet_name='Pairwise_105',
        index=False
    )


    benchmark_info_df.to_excel(
        writer,
        sheet_name='Benchmark_PCA_Cluster',
        index=False
    )


    sorted_distance_df.to_excel(
        writer,
        sheet_name='Raw_Matrix_Sorted'
    )


    sorted_normalized_distance_df.to_excel(
        writer,
        sheet_name='Normalized_Matrix_Sorted'
    )


    summary_df.to_excel(
        writer,
        sheet_name='Summary',
        index=False
    )


print(
    f"\nExcel output: "
    f"{excel_path.resolve()}"
)


# ============================================================

# ============================================================

heatmap_matrix = (
    normalized_distance_matrix[
        np.ix_(
            sort_order,
            sort_order
        )
    ]
)


heatmap_labels = [
    benchmark_compositions[k]
    for k in sort_order
]


heatmap_clusters = (
    benchmark_clusters[
        sort_order
    ]
)


fig, ax = plt.subplots(
    figsize=(13, 11)
)



im = ax.imshow(
    heatmap_matrix,
    aspect='auto',
    vmin=0,
    vmax=1
)


# -------------------------

# -------------------------

ax.set_xticks(
    np.arange(
        len(heatmap_labels)
    )
)

ax.set_yticks(
    np.arange(
        len(heatmap_labels)
    )
)


ax.set_xticklabels(
    heatmap_labels,
    rotation=90,
    fontsize=8
)

ax.set_yticklabels(
    heatmap_labels,
    fontsize=8
)


ax.set_xlabel(
    'Benchmark structures '
    '(sorted by GMM cluster)'
)

ax.set_ylabel(
    'Benchmark structures '
    '(sorted by GMM cluster)'
)


ax.set_title(
    'Normalized pairwise Euclidean distances '
    'among the 15 benchmark structures\n'
    '(PCA space used for the final GMM clustering)'
)


# -------------------------
# Colorbar
# -------------------------

cbar = fig.colorbar(
    im,
    ax=ax
)

cbar.set_label(
    'Normalized Euclidean distance'
)



cbar.set_ticks(
    np.linspace(
        0,
        1,
        6
    )
)


# ============================================================

# ============================================================

for r in range(
    len(heatmap_labels)
):

    for c in range(
        len(heatmap_labels)
    ):

        value = heatmap_matrix[r, c]

        ax.text(
            c,
            r,
            f"{value:.2f}",
            ha='center',
            va='center',
            fontsize=6
        )


# ============================================================

# ============================================================

cluster_boundaries = np.where(
    np.diff(
        heatmap_clusters
    ) != 0
)[0]


for b in cluster_boundaries:

    ax.axhline(
        b + 0.5,
        linewidth=1.2
    )

    ax.axvline(
        b + 0.5,
        linewidth=1.2
    )


# ============================================================

# ============================================================

fig.tight_layout()


figure_path = (
    output_dir
    / '15_benchmark_normalized_euclidean_distance_heatmap.png'
)


fig.savefig(
    figure_path,
    dpi=600,
    bbox_inches='tight'
)


plt.show()


print(
    f"\n："
    f"{figure_path.resolve()}"
)


# ============================================================

# ============================================================

print(
    '\nPairwise distance \u524d10\u4e2a\u6700\u8fd1\u7ed3\u6784\u5bf9\uff1a'
)


display(
    pairwise_df.head(10)
)
```

```python
# ============================================================
# Pairwise Euclidean distance vs FWHM difference
#




#    Figure 1: Normalized Euclidean distance vs |ΔFWHM|
#              Scatter + LOWESS smoothing
#    Figure 2: Distance quartile vs |ΔFWHM|
#              Boxplot + raw points + mean trend

#

#   Normalized_Euclidean_Distance
#   FWHM_Difference_nm
# ============================================================


# ============================================================

# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import spearmanr


# ============================================================

# ============================================================


input_excel = Path(
    r"E:\desktop\Multiobjective Identification of Phosphor Materials via a Domain Knowledge Integrated Semisupervised Learning Framework\code\Density-based clustering\DBSCAN\DBSCAN4\DBSCAN.ipynb"
)





# ============================================================

# ============================================================

figure_dir = Path.cwd() / "FWHM_pair_distance_figures"

figure_dir.mkdir(
    parents=True,
    exist_ok=True
)

print(
    f"：{figure_dir.resolve()}"
)


# ============================================================

# ============================================================

df = pd.read_excel(
    input_excel,
    sheet_name="Pairwise_105"
)

print('\nExcel\u8bfb\u53d6\u6210\u529f\u3002')
print(f"：{len(df)}")

print('\n\u5f53\u524dExcel\u5217\u540d\uff1a')
for col in df.columns:
    print(col)


# ============================================================

# ============================================================

required_columns = [
    "Normalized_Euclidean_Distance",
    "FWHM_Difference_nm"
]

missing_columns = [
    col
    for col in required_columns
    if col not in df.columns
]

if missing_columns:

    raise ValueError(
        f"Excel：{missing_columns}\n"
        '\u8bf7\u68c0\u67e5Excel\u5217\u540d\u662f\u5426\u4e0e\u4ee3\u7801\u5b8c\u5168\u4e00\u81f4\u3002'
    )


# ============================================================

# ============================================================

x = pd.to_numeric(
    df["Normalized_Euclidean_Distance"],
    errors="coerce"
).to_numpy()

y = pd.to_numeric(
    df["FWHM_Difference_nm"],
    errors="coerce"
).to_numpy()



valid_mask = (
    np.isfinite(x)
    &
    np.isfinite(y)
)

x = x[valid_mask]
y = y[valid_mask]


print(
    f"\npair：{len(x)}"
)


# ============================================================

# ============================================================

rho, p_value = spearmanr(
    x,
    y
)

print("\n==============================")
print("Spearman correlation")
print("==============================")

print(
    f"Spearman rho = {rho:.4f}"
)

print(
    f"p-value = {p_value:.4f}"
)


# ============================================================

# ============================================================





use_lowess = False

try:

    from statsmodels.nonparametric.smoothers_lowess import lowess

    lowess_result = lowess(
        endog=y,
        exog=x,


        frac=0.45,


        it=0,

        return_sorted=True
    )

    smooth_x = lowess_result[:, 0]
    smooth_y = lowess_result[:, 1]

    use_lowess = True

    print(
        "\nLOWESS smoothing: statsmodels LOWESS"
    )

except ImportError:

    print(
        '\n\u672a\u68c0\u6d4b\u5230 statsmodels\u3002'
        '\u5c06\u81ea\u52a8\u4f7f\u7528\u79fb\u52a8\u7a97\u53e3\u4e2d\u4f4d\u6570\u8fdb\u884c\u5e73\u6ed1\u3002'
    )

    order = np.argsort(x)

    xs = x[order]
    ys = y[order]


    window = max(
        9,
        len(xs) // 10
    )

    smooth_x = []
    smooth_y = []

    for i in range(
        len(xs)
    ):

        left = max(
            0,
            i - window // 2
        )

        right = min(
            len(xs),
            i + window // 2 + 1
        )

        smooth_x.append(
            xs[i]
        )

        smooth_y.append(
            np.median(
                ys[left:right]
            )
        )

    smooth_x = np.asarray(
        smooth_x
    )

    smooth_y = np.asarray(
        smooth_y
    )


# ============================================================
# 8. Figure 1
# Scatter + LOWESS
# ============================================================

fig, ax = plt.subplots(
    figsize=(6.4, 5.0)
)


# -------------------------

# -------------------------

ax.scatter(
    x,
    y,

    s=38,

    alpha=0.72,

    edgecolors="none"
)


# -------------------------
# LOWESS / smoothing curve
# -------------------------

ax.plot(
    smooth_x,
    smooth_y,

    linewidth=2.2
)


# -------------------------

# -------------------------

ax.set_xlabel(
    "Normalized Euclidean distance",
    fontsize=12
)

ax.set_ylabel(
    r"$|\Delta \mathrm{FWHM}|$ (nm)",
    fontsize=12
)


ax.tick_params(
    axis="both",
    labelsize=10
)


# -------------------------

# -------------------------

if p_value < 0.001:

    p_text = r"$p<0.001$"

else:

    p_text = fr"$p={p_value:.3f}$"


ax.text(
    0.04,
    0.95,

    fr"Spearman $\rho={rho:.3f}$, {p_text}",

    transform=ax.transAxes,

    fontsize=11,

    va="top"
)


# -------------------------

# -------------------------

ax.spines[
    "top"
].set_visible(
    False
)

ax.spines[
    "right"
].set_visible(
    False
)


fig.tight_layout()


# ============================================================

# ============================================================

svg_path_1 = (
    figure_dir
    /
    "FWHM_distance_scatter_LOWESS.svg"
)

png_path_1 = (
    figure_dir
    /
    "FWHM_distance_scatter_LOWESS.png"
)


fig.savefig(
    svg_path_1,

    format="svg",

    bbox_inches="tight"
)


fig.savefig(
    png_path_1,

    dpi=600,

    bbox_inches="tight"
)


plt.show()


print(
    f"\nFigure 1 SVG：{svg_path_1.resolve()}"
)

print(
    f"Figure 1 PNG：{png_path_1.resolve()}"
)


# ============================================================

# ============================================================


sort_idx = np.argsort(x)

n_samples = len(x)





rank_group = np.zeros(
    n_samples,
    dtype=int
)


for rank, original_index in enumerate(
    sort_idx
):

    group = int(
        rank * 4 / n_samples
    )

    group = min(
        group,
        3
    )

    rank_group[
        original_index
    ] = group


# Q1-Q4
groups = [

    y[
        rank_group == i
    ]

    for i in range(
        4
    )
]



distance_groups = [

    x[
        rank_group == i
    ]

    for i in range(
        4
    )
]


# ============================================================

# ============================================================

group_means = [

    np.mean(
        values
    )

    for values in groups
]


group_medians = [

    np.median(
        values
    )

    for values in groups
]


print(
    "\n=========================================="
)

print(
    "Distance quartile statistics"
)

print(
    "=========================================="
)


for i in range(
    4
):

    values = groups[i]

    distances = distance_groups[i]

    print(
        f"\nQ{i+1}:"
    )

    print(
        f"n = {len(values)}"
    )

    print(
        "Distance range = "
        f"{distances.min():.4f} – "
        f"{distances.max():.4f}"
    )

    print(
        "Mean |ΔFWHM| = "
        f"{np.mean(values):.2f} nm"
    )

    print(
        "Median |ΔFWHM| = "
        f"{np.median(values):.2f} nm"
    )


# ============================================================
# 12. Figure 2
# Distance quartile boxplot
# ============================================================

fig, ax = plt.subplots(
    figsize=(6.4, 5.0)
)


# -------------------------
# Boxplot
# -------------------------

boxplot_result = ax.boxplot(
    groups,

    positions=[
        1,
        2,
        3,
        4
    ],

    widths=0.52,

    showfliers=False,

    patch_artist=False,

    medianprops={
        "linewidth": 1.8
    }
)


# ============================================================

# ============================================================




rng = np.random.default_rng(
    42
)


for position, values in enumerate(
    groups,
    start=1
):

    jitter = rng.normal(
        loc=0,
        scale=0.055,
        size=len(values)
    )

    x_points = (
        np.full(
            len(values),
            position,
            dtype=float
        )
        +
        jitter
    )

    ax.scatter(
        x_points,
        values,

        s=28,

        alpha=0.65,

        edgecolors="none"
    )


# ============================================================

# ============================================================

ax.plot(
    [
        1,
        2,
        3,
        4
    ],

    group_means,

    marker="o",

    markersize=5,

    linewidth=1.7,

    label="Mean"
)


# ============================================================

# ============================================================

ax.set_xticks(
    [
        1,
        2,
        3,
        4
    ]
)


ax.set_xticklabels(
    [
        "Q1\nNearest",
        "Q2",
        "Q3",
        "Q4\nFarthest"
    ],

    fontsize=10
)


ax.set_xlabel(
    "Quartile of normalized Euclidean distance",
    fontsize=12
)


ax.set_ylabel(
    r"$|\Delta \mathrm{FWHM}|$ (nm)",
    fontsize=12
)


ax.tick_params(
    axis="y",
    labelsize=10
)


# -------------------------
# Legend
# -------------------------

ax.legend(
    frameon=False,

    fontsize=10
)


# -------------------------

# -------------------------

ax.spines[
    "top"
].set_visible(
    False
)

ax.spines[
    "right"
].set_visible(
    False
)


fig.tight_layout()


# ============================================================

# ============================================================

svg_path_2 = (
    figure_dir
    /
    "FWHM_distance_quartile_boxplot.svg"
)

png_path_2 = (
    figure_dir
    /
    "FWHM_distance_quartile_boxplot.png"
)


fig.savefig(
    svg_path_2,

    format="svg",

    bbox_inches="tight"
)


fig.savefig(
    png_path_2,

    dpi=600,

    bbox_inches="tight"
)


plt.show()


print(
    f"\nFigure 2 SVG：{svg_path_2.resolve()}"
)

print(
    f"Figure 2 PNG：{png_path_2.resolve()}"
)


# ============================================================

# ============================================================

print(
    "\n=========================================="
)

print(
    "Final summary"
)

print(
    "=========================================="
)


print(
    f"Number of pairs = {len(x)}"
)


print(
    f"Spearman rho = {rho:.4f}"
)


print(
    f"Spearman p = {p_value:.4f}"
)


for i in range(
    4
):

    print(
        f"Q{i+1}: "
        f"mean |ΔFWHM| = "
        f"{group_means[i]:.2f} nm, "
        f"median = "
        f"{group_medians[i]:.2f} nm"
    )
```
